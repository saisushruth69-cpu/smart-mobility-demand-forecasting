```python
data = data.drop_duplicates()

print("Shape after removing duplicates:")
print(data.shape)
```

```python
data = data.drop(
    ['instant', 'casual', 'registered'],
    axis=1
)

print(data.head())
```

```python
data['dteday'] = pd.to_datetime(data['dteday'])

print(data['dteday'].dtype)
```

```python
data = data.sort_values(
    by=['dteday', 'hr']
).reset_index(drop=True)
```

```python
features = [
    'season',
    'yr',
    'mnth',
    'hr',
    'holiday',
    'weekday',
    'workingday',
    'weathersit',
    'temp',
    'atemp',
    'hum',
    'windspeed'
]

X = data[features]

print(X.head())
```

```python
y = data['cnt']

print(y.head())
```

```python
categorical_columns = [
    'season',
    'mnth',
    'hr',
    'holiday',
    'weekday',
    'workingday',
    'weathersit'
]

X = pd.get_dummies(
    X,
    columns=categorical_columns,
    drop_first=True,
    dtype=int
)

print(X.head())
print("Total features:", X.shape[1])
```

```python
split_index = int(len(X) * 0.8)

X_train = X.iloc[:split_index]
X_test = X.iloc[split_index:]

y_train = y.iloc[:split_index]
y_test = y.iloc[split_index:]

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)

print("y_train:", y_train.shape)
print("y_test:", y_test.shape)

print("Missing values:", X_train.isnull().sum().sum())
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import numpy as np
import pandas as pd
```

```python
# Create model
lr_model = LinearRegression()

# Train model
lr_model.fit(X_train, y_train)

# Predict demand
lr_predictions = lr_model.predict(X_test)

print("Linear Regression completed!")
```

```python
# Create model
dt_model = DecisionTreeRegressor(
    max_depth=12,
    random_state=42
)

# Train model
dt_model.fit(X_train, y_train)

# Predict demand
dt_predictions = dt_model.predict(X_test)

print("Decision Tree completed!")
```

```python
# Create model
rf_model = RandomForestRegressor(
    n_estimators=100,
    max_depth=15,
    random_state=42,
    n_jobs=-1
)

# Train model
rf_model.fit(X_train, y_train)

# Predict demand
rf_predictions = rf_model.predict(X_test)

print("Random Forest completed!")
```

```python
def evaluate_model(name, actual, predicted):

    mae = mean_absolute_error(actual, predicted)

    mse = mean_squared_error(actual, predicted)

    rmse = np.sqrt(mse)

    r2 = r2_score(actual, predicted)

    return {
        'Model': name,
        'MAE': mae,
        'MSE': mse,
        'RMSE': rmse,
        'R2 Score': r2
    }


results = []

results.append(
    evaluate_model("Linear Regression",
                   y_test, lr_predictions)
)

results.append(
    evaluate_model("Decision Tree",
                   y_test, dt_predictions)
)

results.append(
    evaluate_model("Random Forest",
                   y_test, rf_predictions)
)

results_df = pd.DataFrame(results)

results_df
```

## Demand Profile Clustering

K-Means clustering is used to group bicycle usage conditions into Low, Medium, and High demand profiles.

Random Forest is used for numerical bicycle demand prediction. K-Means is used for demand profiling, not for predicting the exact number of bicycles.


```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
cluster_features = [
    'season',
    'hr',
    'workingday',
    'weathersit',
    'temp',
    'atemp',
    'hum',
    'windspeed'
]

cluster_data = data[cluster_features]

cluster_data.head()
```

```python
cluster_train = cluster_data.iloc[:split_index]

cluster_test = cluster_data.iloc[split_index:]
```

```python
scaler = StandardScaler()

scaled_train = scaler.fit_transform(cluster_train)

scaled_test = scaler.transform(cluster_test)

print("Scaling completed!")
```

```python
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt

inertia = []
silhouette_scores = []

k_values = range(2, 8)

for k in k_values:

    km = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = km.fit_predict(scaled_train)

    inertia.append(km.inertia_)

    silhouette_scores.append(
        silhouette_score(
            scaled_train,
            labels
        )
    )

# Elbow Method
plt.figure(figsize=(8,5))

plt.plot(
    list(k_values),
    inertia,
    marker='o'
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method for K-Means")

plt.show()


# Silhouette Score
plt.figure(figsize=(8,5))

plt.plot(
    list(k_values),
    silhouette_scores,
    marker='o'
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score for K-Means")

plt.show()
```

```python
kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

train_clusters = kmeans.fit_predict(scaled_train)

test_clusters = kmeans.predict(scaled_test)

print("Clustering completed!")
```

```python
train_data = data.iloc[:split_index].copy()

train_data['Cluster'] = train_clusters

cluster_summary = train_data.groupby('Cluster')['cnt'].agg(
   ['mean', 'min', 'max', 'count']
)

cluster_summary
```

```python
cluster_order = cluster_summary['mean'].sort_values()

demand_names = ['Low Demand', 'Medium Demand', 'High Demand']

cluster_mapping = {}

for i, cluster_id in enumerate(cluster_order.index):
    cluster_mapping[cluster_id] = demand_names[i]

train_data['Demand_Profile'] = train_data['Cluster'].map(cluster_mapping)

print(train_data[['hr', 'cnt', 'Cluster', 'Demand_Profile']].head(15))
```

```python
plt.figure(figsize=(10, 6))

for cluster_id in range(3):

    cluster_points = train_data[
        train_data['Cluster'] == cluster_id
    ]

    plt.scatter(
        cluster_points['hr'],
        cluster_points['cnt'],
        label=cluster_mapping[cluster_id],
        alpha=0.5
    )

plt.title("Bicycle Demand Clusters")
plt.xlabel("Hour of the Day")
plt.ylabel("Bicycle Rentals")

plt.legend()
plt.grid()

plt.show()
```

```python
test_data = data.iloc[split_index:].copy()

test_data['Cluster'] = test_clusters

test_data['Demand_Profile'] = test_data['Cluster'].map(cluster_mapping)

test_data[['hr', 'cnt', 'Cluster', 'Demand_Profile']].head(15)
 
```

```python
clustering_package = {
    'scaler': scaler,
    'kmeans': kmeans,
    'features': cluster_features,
    'cluster_mapping': cluster_mapping
}

joblib.dump(
    clustering_package,
    'bike_clustering_model.pkl'
)

print("Clustering model saved successfully!")
```

```python
capacity_data = test_data.copy()

# Use predictions from the selected Random Forest model
rf_predictions = best_model.predict(X_test)

capacity_data['Predicted_Demand'] = (
    np.maximum(rf_predictions, 0)
)

capacity_data['Actual_Demand'] = y_test.values

capacity_data['Demand_Profile'] = (
    capacity_data['Cluster']
    .map(cluster_mapping)
)

capacity_data[
    [
        'hr',
        'Predicted_Demand',
        'Actual_Demand',
        'Demand_Profile'
    ]
].head(10)
```

```python
# Operational inputs
fleet_capacity = 500
current_available_bikes = 400

# Safety buffer
safety_buffer = 0.10

capacity_data['Required_Bikes'] = np.ceil(
    capacity_data['Predicted_Demand']
    * (1 + safety_buffer)
).astype(int)

capacity_data['Shortage'] = np.maximum(
    capacity_data['Required_Bikes']
    - current_available_bikes,
    0
)

capacity_data['Surplus'] = np.maximum(
    current_available_bikes
    - capacity_data['Required_Bikes'],
    0
)

capacity_data['Utilization_%'] = (
    capacity_data['Predicted_Demand']
    / fleet_capacity
) * 100

capacity_data.head(10)
```

```python
def capacity_recommendation(row):

    required = row['Required_Bikes']
    available = current_available_bikes
    fleet = fleet_capacity

    if required > fleet:

        return "Fleet Capacity Insufficient"

    elif required > available:

        return "Increase / Reallocate Bikes"

    elif required < available * 0.70:

        return "Reduce / Reallocate Surplus"

    else:

        return "Maintain Availability"


capacity_data['Recommendation'] = (
    capacity_data.apply(
        capacity_recommendation,
        axis=1
    )
)

capacity_data[
    [
        'hr',
        'Predicted_Demand',
        'Required_Bikes',
        'Shortage',
        'Surplus',
        'Utilization_%',
        'Demand_Profile',
        'Recommendation'
    ]
].head(15)
```

```python
capacity_data[
    [
        'hr',
        'Predicted_Demand',
        'Required_Bikes',
        'Utilization_%',
        'Recommendation'
    ]
].sort_values(
    by='Predicted_Demand',
    ascending=False
).head(10)
```

```python
capacity_report = capacity_data.groupby(
    'Recommendation'
).agg(
    Average_Demand=(
        'Predicted_Demand',
        'mean'
    ),

    Maximum_Demand=(
        'Predicted_Demand',
        'max'
    ),

    Average_Utilization=(
        'Utilization_%',
        'mean'
    ),

    Total_Shortage=(
        'Shortage',
        'sum'
    ),

    Total_Surplus=(
        'Surplus',
        'sum'
    ),

    Total_Hours=(
        'hr',
        'count'
    )
)

capacity_report
```

```python
# ==========================================================
# PEAK DEMAND IDENTIFICATION & HOURLY ANALYSIS
# ==========================================================
# NOTE: the dataframe created in the previous cells is 'capacity_data'
# (there is no variable called 'df' in this notebook).

peak_df = capacity_data.copy()

pred_col = 'Predicted_Demand'
hour_col = 'hr'

# 1. Mark the top 10% of predicted demand as Peak (90th percentile threshold)
peak_threshold = peak_df[pred_col].quantile(0.90)
peak_df['Is_Peak'] = (peak_df[pred_col] >= peak_threshold).astype(int)

print('--- Peak Demand Cutoff ---')
print(f'90th Percentile Threshold: {peak_threshold:.2f}')
print(f"Total Peak Records Identified: {peak_df['Is_Peak'].sum()} out of {len(peak_df)} rows ({peak_df['Is_Peak'].mean()*100:.1f}%)")

# 2. Analyze Peak Hours (distribution of peak demand across the day)
print('\n--- Peak Hours Analysis ---')
peak_hour_analysis = (
    peak_df.groupby(hour_col)
    .agg(
        Total_Count=(hour_col, 'count'),
        Peak_Occurrences=('Is_Peak', 'sum'),
        Avg_Predicted_Demand=(pred_col, 'mean')
    )
    .assign(Peak_Frequency_Pct=lambda x: (x['Peak_Occurrences'] / x['Total_Count'] * 100).round(1))
    .sort_values(by='Peak_Occurrences', ascending=False)
)

print(peak_hour_analysis)

# Most frequent peak hours
top_peak_hours = peak_hour_analysis[peak_hour_analysis['Peak_Occurrences'] > 0].index.tolist()
print(f'\nIdentified Peak Hours (ranked by frequency): {top_peak_hours}')

# 3. Save the CSV with the 'Is_Peak' column
output_csv = 'predicted_demand_with_peaks.csv'
peak_df.to_csv(output_csv, index=False)
print(f'\n[OK] CSV successfully saved with Is_Peak column to: {output_csv}')

# Keep capacity_data in sync
capacity_data['Is_Peak'] = peak_df['Is_Peak']
```
